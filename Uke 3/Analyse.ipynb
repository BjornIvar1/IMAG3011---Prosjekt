# Samlet Analyse

I dette prosjektet har jeg eksperimentert med hvordan likeverdige beskrivelser av det samme polynomet kan gi forskjellige numeriske resultater. Gjennom prosjektet har jeg sammenlignet polynomer skrevet i to forskjellige basiser, monomialbasis og Chebyshevbasis.

Det siste eksperimentet i $T_{50}$ ga den største absolutt forskjellen. Her ga monomialevalueringen en verdi på omtrent $-31.59$, mens cosinusreferansen var på omtrent $0.70$. Dette skapte en feil på $32.29$. På den andre siden så oppnådde Chebyshev et svært bedre resultatet og hadde en feil på $2.33e-15$. Dette viser at to beskrivelser av et polynomrom kan gi to svært ulike numeriske resultater.

Polynomrommet $P_n$ består av alle polynomer med grad n, mens koordinatene er koeffisientene polynomet får i en basis. Det samme polynomet kan derfor ha forskjellige koordinater i monomialbasis og Chebyshevbasis. Selv om koeffisientlistene er forskjellige, representerer de fortsatt det samme matematiske polynomet.

Avlesningsavbildningen $E$ tar inn et polynom og returnerer polynomverdiene i de valgte avlesningspunktene. Matrisene $M$ og $C$ representerer den samme avbildningen i henholdsvis monomialbasis og Chebyshevbasis. I basisforsøket ble referansepolynomet, avlesningspunktene og avlesningsverdiene holdt fast, mens bare basisen ble endret. I punktplasseringseksperimentet ble derimot Chebyshevbasis holdt fast, mens avlesningspunktene ble flyttet. Derfor undersøkte de to forsøkene forskjellige årsaker til numeriske feil.

De største feilene oppstod i intervallet nær grenseområdene $[-1, 1]$. Her ser vi i de jevnt fordelte punktene at det oppstod store feil i de to områdene. I de cosinusfordelte punktene så lå punktene tettere sammen i de to områdene, noe som var med på å redusere feilen.

Ved grad 30 ble monomialkoeffisienten på opptil $3.2e7$ og i $T_{50}$-forsøket ble den største koeffisienten $1.29e18$. Når vi jobber med så store tall og verdien i $T_{50}(x)$ skal ligge mellom -1 og 1, dermed må monomialbasisen kansellere hverandre ut for å ende opp i det intervallet. Når vi jobber i Python så vil regneoperasjoner med store tall gi avrundingsfeil, som blir med på å skape feil.

Feilpolynomet $r$ var nesten usynlig for avlesningsavbildningen. Ved avlesningspunktene hadde det bare verdiene $10^{-10}$, slik at $E(r)$ var svært nær null. Mellom de jevnt fordelte punktene vokste likevel $|r(x)|$ til omtrent $6.601e-4$. Dette er numerisk nesten-informasjonstap. Nullromspolynomet $z$ fra del 1 var derimot helt usynlig for avlesningsavbildningen, fordi $E(z)=0$ nøyaktig.

Matematisk viste jeg at $T_0,\ldots,T_4$ danner en basis for $P_4$, og at fire forskjellige avlesningspunkter bestemmer et polynom i $P_3$ entydig. Jeg viste også at $E(z)=0$ ved direkte innsetting. De konkrete feilverdiene, maksimumspunktene og forsterkningene bygger derimot på numeriske forsøk og kontroll på et endelig rutenett.

Ved grad 3 ga begge basisene samme feil på omtrent $4.44e-16$. Her var det vanskelig å se forskjellen i numerisk kvalitet når graden var lav. Senere i $T_{50}$-eksperimentet så vokste monomialkoeffisientene til $1.29e18$, og monomialevalueringen fikk en feil på omtrent $32.3$. Mens Chebyshevevalueringen hadde en svært liten feil på omtrent $2.33e-15$. Forsøket viste her at monomialbasis er svak for avrundingsfeil og feilen kan bli større når graden økes, selv om basisen representerer samme polynome.
