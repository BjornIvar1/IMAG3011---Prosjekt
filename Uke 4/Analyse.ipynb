# Samlet analyse

I dette prosjektet har jeg undersøkt hvordan støy, basisvalg og beregningsmetode påvirker en tilpasning. Gjennom prosjektet har jeg sammenlignet residualer, QR-kontroller og kurvefeil for å undersøke om tilpasningene er pålitelige.

#### Når skyldes avviket at modellen ikke passer målingene, og når tyder kontrollene på problemer i beregningen?

I del 1 var residualen til QR-modellen ikke null, som betyr at linjen ikke treffer alle målingene. Samtidig var QR-kontrollene nær null, og koeffisientene stemte med `lstsq`. Dette viser at en residual som ikke er null, ikke nødvendigvis betyr at beregningen er feil.

I støyforsøket økte residualnormen fra omtrent $0.002536$ til $0.02536$ når støyen økte fra $10^{-3}$ til $10^{-2}$. Kontrollene var uendret fordi matrisen var den samme. Her skyldtes økningen i avviket støyen i målingene.

For klassisk GS i monomialbasis ved grad 20 var ortogonalitetsfeilen $1.454$, og kurvefeilen var $1.246\cdot10^{-2}$. På samme matrise og målinger ga `lstsq` en kurvefeil på $6.362\cdot10^{-10}$. Her ser vi at avviket også skyldes problemer i beregningen.

#### Hva forteller residual, ortogonalitet og kurvefeil hver for seg?

Residualen viser forskjellen mellom målingene og modellverdiene. En liten residualnorm betyr at modellen passer godt til målingene, men den kan fortsatt avvike fra referansekurven som er uten støy. RMSE viser avvik per måling og gjør det lettere å sammenligne forsøk med forskjellig antall målinger.

Ortogonalitetsfeilen undersøker om kolonnene i $Q$ har lengde 1 og står vinkelrett på hverandre. Kurvefeilen viser avviket mellom den tilpassede kurven og referansekurven, også mellom målepunktene.

Ved grad 20 var den relative faktoriseringsfeilen for klassisk GS bare $7.649\cdot10^{-17}$. Produktet $QR$ var derfor svært nær $A$, men ortogonalitetsfeilen var stor. Forsøket viser at en liten faktoriseringsfeil alene ikke er nok til å stole på tilpasningen.

#### Hva endret du i redningsforsøket, og hvilke resultater støtter vurderingen?

Jeg byttet fra klassisk GS til modifisert GS. Referansepolynomet, graden, målepunktene og målingene ble holdt like, slik at bare metoden ble endret.

Residualnormen sank fra $7.163\cdot10^{-3}$ til $5.339\cdot10^{-10}$, og RMSE sank fra $1.092\cdot10^{-3}$ til $8.143\cdot10^{-11}$. Den største kurvefeilen sank også fra $1.246\cdot10^{-2}$ til $9.069\cdot10^{-10}$. Tilpasningen kom derfor mye nærmere både målingene og referansekurven.

Ortogonalitetsfeilen sank fra $1.454$ til $5.444\cdot10^{-10}$, som viser at kolonnene i $Q$ ble nærmere ortonormale. Den relative faktoriseringsfeilen var svært liten for begge metodene. Modifisert GS ga dermed en tydelig forbedring i dette forsøket.

#### Hvilken begrensning ved forsøket gjør at du bør være forsiktig med å generalisere?

Jeg har bare undersøkt et begrenset antall grader, målepunkter og støymønstre. Redningsforsøket brukte grad 20 og ett fast frø. Derfor kan jeg ikke konkludere med at én metode eller basis alltid gir den beste kurven.

I forsøket med flere målinger var konklusjonen heller ikke den samme for alle frøene. For frø 2026 sank kurvefeilen fra omtrent $0.000798$ til $0.000394$. For frø 2027 sank den ved 16 målinger, men økte igjen ved 32. Flere målinger betyr derfor ikke at kurvefeilen alltid blir mindre.

Kurvefeilen ble også undersøkt på et endelig rutenett. Resultatene viser feilen ved kontrollpunktene, men gir ikke et matematisk bevis for hvor stor feilen er i alle punktene mellom dem.
